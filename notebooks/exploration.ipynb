# Myntra Review Scrapper — Exploratory Analysis

Quick ad-hoc analysis of a scraped reviews CSV (e.g. `data/latest_reviews.csv`).
Run `app.py` or `streamlit_app.py` first to generate a CSV, then explore it here.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
from scraper import scrape_reviews

pd.set_option("display.max_colwidth", 80)
```

```python
product_url = "https://www.myntra.com/tshirts/roadster/sample-product/123456/buy"
df = scrape_reviews(product_url)
df.head()
```

```python
df["rating"].describe()
```

```python
df["rating"].value_counts().sort_index().plot(kind="bar", title="Rating Distribution")
```

```python
df["review_length"] = df["review_text"].str.len()
df[["review_text", "review_length", "rating"]].sort_values("review_length", ascending=False).head()
```
